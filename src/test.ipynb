```python
import matplotlib.pyplot as plt
import fastf1

def practice_drivers_comparison(year, gp, id):
   # Loading the GP
   session = fastf1.get_session(year, gp, id)
   session.load()

   # Creating a DF with the quick laps & a list of teams to be used as filters later  
   df = session.laps.pick_drivers(session.drivers).pick_quicklaps()
   df["LapTime(s)"] = df["LapTime"].dt.total_seconds()

   # **Sort teams by their FASTEST lap time**
   team_order = df.groupby("Team")["LapTime(s)"].min().sort_values().index.tolist()
#    df_teams = df['Team'].drop_duplicates()
#    teams = [team for team in df_teams]

   # **Find Global Min & Max Lap Time**
   min_time = df["LapTime(s)"].min()
   max_time = df["LapTime(s)"].max()

   # **Apply Buffer ±1 sec**
   min_time -= 1
   max_time += 1

   # Create a single figure with 2 rows and 5 columns
   fig, axs = plt.subplots(nrows=2, ncols=5, figsize=(20, 10))
   fig.suptitle(f'{session} - Team Comparison', fontsize=18)  
   axs = axs.flatten()  # Flatten the grid to iterate easily over subplots
   first_legend = None  # Placeholder to store the first legend
   fig.patch.set_facecolor('#283747')
#    axs.set_facecolor('#283747')

   # Creating new DFs per team, to be used for the charts
   for i, team in enumerate(team_order):
       ax = axs[i]  # Select subplot for the current team
       ax.set_facecolor('#283747')
       team_df = df[df['Team'] == team]

       # Boxplot for lap times per driver
       sns.boxplot(
           data=team_df, 
           x="Driver", 
           y="LapTime(s)", 
           hue="Driver", 
           palette=fastf1.plotting.get_driver_color_mapping(session=session), 
           ax=ax
       )

       # Swarmplot for tire compounds
       swarm = sns.swarmplot(
           data=team_df, 
           x="Driver", 
           y="LapTime(s)", 
           hue="Compound", 
           palette=fastf1.plotting.get_compound_mapping(session=session), 
           hue_order=["SOFT", "MEDIUM", "HARD"], 
           linewidth=0, 
           size=4, 
           ax=ax
       )

       # **Set consistent y-axis limits**
       ax.set_ylim(min_time, max_time)

       # Store the first legend and remove others
       if first_legend is None:
           first_legend = ax.legend(loc='upper right', title="Compound")  # Store first legend
       else:
           ax.legend_.remove()  # Remove legends from other subplots


   # Hide unused subplots if fewer than 10 teams
   for i in range(len(teams), len(axs)):
       fig.delaxes(axs[i])

   plt.tight_layout()  # Adjust layout
   plt.show()
```

```python
import fastf1.plotting as f1plot
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Setup FastF1 plot style
f1plot.setup_mpl(mpl_timedelta_support=True, misc_mpl_mods=False)

fig, ax = plt.subplots(figsize=(20, 8))
fig.patch.set_facecolor('#2c3e50')
ax.set_facecolor('#2c3e50') 

# Get color mapping per driver
driver_colors = f1plot.get_driver_color_mapping(session=session)

# Line styles: solid for faster, dotted for slower
style_map = ['solid', 'dotted']

# Get first 5 unique drivers (to control total count shown)
top_5_drivers = driver_laps["Driver"].unique()

# Filter data for only the first 5 drivers
filtered_laps = driver_laps[driver_laps["Driver"].isin(top_5_drivers)]

# Group by team from selected drivers
teams = filtered_laps['Team'].unique()

for team in teams:
    team_df = filtered_laps[filtered_laps["Team"] == team]
    team_drivers = team_df["Driver"].unique()

    # Compute average lap time for each driver
    driver_avg = (
        team_df.groupby("Driver")["LapTime"]
        .mean()
        .sort_values()
    )

    # Assign line styles based on speed: faster = solid, slower = dotted
    for i, (driver, _) in enumerate(driver_avg.items()):
        driver_df = team_df[team_df["Driver"] == driver]

        sns.lineplot(
            data=driver_df,
            x="LapNumber",
            y="LapTime",
            ax=ax,
            label=driver,
            color=driver_colors.get(driver),
            linewidth=2,
            linestyle=style_map[min(i, 1)]  # 0=solid, 1=dotted
        )

# Label axes
ax.set_xlabel("Lap Number")
ax.set_ylabel("Lap Time")

# Limit y-axis to show only lap times between 1:35.0 and 1:42.0
# ax.set_ylim(pd.to_timedelta("0:01:40.0"), pd.to_timedelta("0:01:35.0"))

# Invert y-axis so faster times are higher
ax.invert_yaxis()

# Set x-axis ticks every 5 laps
ax.set_xticks(range(0, int(driver_laps["LapNumber"].max()) + 1, 2))

# Grid and styling
plt.grid(color='w', which='major', axis='both', alpha=0.5)
sns.despine(left=True, bottom=True)

plt.tight_layout()
plt.show()
```

```python
cleaned_laps = driver_laps.copy()

cleaned_laps['LapTime(s)'] = cleaned_laps['LapTime'].dt.total_seconds()

def replace_outliers_time_with_mean(group): 
    q1 = group['LapTime(s)'].quantile(0.25)
    q3 = group['LapTime(s)'].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    mean_time = group["LapTime(s)"].mean()

    group["LapTime_Cleaned"] = group["LapTime(s)"].apply(
        lambda x: mean_time if x < lower_bound or x > upper_bound else x
    )
    return group

# Apply the function per driver
cleaned_laps = cleaned_laps.groupby("Driver").apply(replace_outliers_time_with_mean)
```

```python
year = 2025
gp = 'China'
id = 'S'

session = fastf1.get_session(year, gp, id)
session.load()


fastest_lap = session.laps.pick_fastest()
car_data = fastest_lap.get_car_data().add_distance()

circuit_info = session.get_circuit_info()

team_color = fastf1.plotting.get_team_color(fastest_lap['Team'],
                                            session=session)

fig, ax = plt.subplots()
ax.plot(car_data['Distance'], car_data['Speed'],
        color=team_color, label=fastest_lap['Driver'])

# Draw vertical dotted lines at each corner that range from slightly below the
# minimum speed to slightly above the maximum speed.
v_min = car_data['Speed'].min()
v_max = car_data['Speed'].max()
ax.vlines(x=circuit_info.corners['Distance'], ymin=v_min-20, ymax=v_max+20,
          linestyles='dotted', colors='grey', alpha=0.7)

# Plot the corner number just below each vertical line.
# For corners that are very close together, the text may overlap. A more
# complicated approach would be necessary to reliably prevent this.
for _, corner in circuit_info.corners.iterrows():
    txt = f"{corner['Number']}{corner['Letter']}"
    ax.text(corner['Distance'], v_min-30, txt,
            va='center_baseline', ha='center', size='small')

ax.set_xlabel('Distance in m')
ax.set_ylabel('Speed in km/h')
ax.legend()

# Manually adjust the y-axis limits to include the corner numbers, because
# Matplotlib does not automatically account for text that was manually added.
ax.set_ylim([v_min - 40, v_max + 20])

plt.show()
```
